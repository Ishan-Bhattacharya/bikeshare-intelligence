```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
ROOT = Path.cwd().parent

RAW_DATA_PATH = ROOT / "data"/"raw"/"hour.csv"
PROCESSED_DATA_PATH = ROOT / "data"/"processed"

raw_data = pd.read_csv(RAW_DATA_PATH)

cleaned_data = raw_data.copy()

print(f"Raw data shape: {raw_data.shape}")
print(f"Cleaned data shape: {cleaned_data.shape}")

cleaned_data.head()
```

```python
print("Dataset shape:", cleaned_data.shape)
print("Dataset columns:", cleaned_data.columns.tolist())

print("Data Types:\n", cleaned_data.dtypes)
print("Data Info:")
cleaned_data.info()
```

```python
## Converting 'dteday' column to datetime format
print("Before conversion, 'dteday' column data type:", cleaned_data['dteday'].dtype)


cleaned_data['dteday'] = pd.to_datetime(cleaned_data['dteday'], format='%Y-%m-%d', errors='raise')

print("After conversion, 'dteday' column data type:", cleaned_data['dteday'].dtype)

cleaned_data['dteday'].head()
```

```python
cleaned_data.dtypes  #Verifying the conversion of the 'dteday' column to datetime format
```

```python
print("Datatype:", cleaned_data['dteday'].dtype)

print(
    "Missing dates:",
    cleaned_data['dteday'].isna().sum()
)

print(
    "Date range:",
    cleaned_data['dteday'].min(),
    "to",
    cleaned_data['dteday'].max()
)
```

```python
#Post-cleaning validation checks

print("Original shape:", raw_data.shape)
print("Cleaned shape:", cleaned_data.shape)

print(
    "Rental count mismatches:",
    (
        cleaned_data['cnt'] !=
        cleaned_data['casual'] + cleaned_data['registered']
    ).sum()
)

print(
    "Missing values:",
    cleaned_data.isnull().sum().sum()
)

print(
    "Duplicate records:",
    cleaned_data.duplicated().sum()
)
```

```python
PROCESSED_DATA_PATH.mkdir(parents=True, exist_ok=True)

OUTPUT_PATH = PROCESSED_DATA_PATH / "hourly_cleaned.parquet"
OUTPUT_PATH_CSV = PROCESSED_DATA_PATH / "hourly_cleaned.csv"

cleaned_data.to_parquet(OUTPUT_PATH, index=False)
cleaned_data.to_csv(OUTPUT_PATH_CSV, index=False)

print("Cleaned dataset saved to:", OUTPUT_PATH) 
print("Cleaned dataset saved to:", OUTPUT_PATH_CSV)
```

## Summary

The `dteday` column in the dataset was originally in string format and needed cleaning.
In this notebook the following are done
- The `dteday` column has been converted to `datetime64[us]` format
- The format has been set in `%YYYY-%MM-%DD`
- The cleaned data has been exported in the processed folder
