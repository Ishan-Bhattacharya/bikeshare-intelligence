```python
#Frequency counts for categorical columns

categorical_columns = [
    'season',
    'yr',
    'mnth',
    'hr',
    'holiday',
    'weekday',
    'workingday',
    'weathersit'
]

for col in categorical_columns:
    print(f"\nColumn: {col}")
    print(hourly_data[col].value_counts(dropna=False).sort_index())
```

```python
#Convert the 'dteday' column to datetime format for easier analysis

dates = pd.to_datetime(hourly_data['dteday'])

print(f"Start Date: {dates.min()}")
print(f"End Date: {dates.max()}")

print(f"Total Days: {(dates.max() - dates.min()).days + 1}")
print(f"Unique Days: {dates.nunique()}")
print(f"Missing Days: {(dates.max() - dates.min()).days + 1 - dates.nunique()}")
print(f"Total Hours: {hourly_data['hr'].nunique()}")
```

```python
# Check for starting and ending timestamps for the dataset by combining the date and hour columns to create a timestamp for each hour in the dataset.

timestamps =(
    dates + pd.to_timedelta(hourly_data['hr'], unit='h') # Create timestamp for each hour
)

print(f"Start Timestamp: {timestamps.min()}")
print(f"End Timestamp: {timestamps.max()}")
```

```python
# Check for duplicate records in the dataset based on the combination of date and hour, as well as duplicate rows in general.

print(f"Duplicate date-hourly records: {hourly_data.duplicated(subset=['dteday', 'hr']).sum()}")

print(f"Duplicate rows: {hourly_data.duplicated().sum()}")
```

```python
#Checking if the time series is complete by comparing the expected range of timestamps with the observed timestamps in the dataset.

expected_range = pd.date_range(start=timestamps.min(), end=timestamps.max(), freq='h')


observed_hours = pd.DatetimeIndex(pd.to_datetime(timestamps.unique()))
missing_hours = expected_range.difference(observed_hours)


print(f"Expected number of hours: {len(expected_range)}")
print(f"Observed number of hours: {len(observed_hours)}")
print(f"Missing hours: {len(missing_hours)}")

print("Missing hours:")
for hour in missing_hours:
    print(hour)
```

```python
completeness = (
    len(observed_hours) / len(expected_range)
) * 100

missing_percentage = (
    len(missing_hours) / len(expected_range)
) * 100

print(f"Completeness: {completeness:.2f}%")
print(f"Missing Percentage: {missing_percentage:.2f}%")  #Shows missing percentage of hours in the dataset which makes it easier to understand the extent of missing data in the dataset.
```

```python
#Check if these missing values are random or in some specific pattern. For example, if the missing values are concentrated in a specific month or season, it may indicate a systematic issue with data collection during that time period.

missing_by_hour = pd.Series(
    missing_hours.hour
).value_counts().sort_index()

print(f"Missing by Hour:\n{missing_by_hour}")

```

```python
#Chwcking for consecutive missing hours to see if there are any long gaps in the data that may indicate a more serious issue with data collection.
missing_df = pd.DataFrame({
    'timestamp': missing_hours
})

missing_df['gap_group'] = (    #Make a column that groups consecutive missing hours together by checking if the difference between each timestamp and the previous one is not equal to 1 hour. If it is not equal to 1 hour, it means there is a gap in the data, and we increment the gap group number.
    missing_df['timestamp']
    .diff()
    .ne(pd.Timedelta(hours=1))
    .cumsum()
)

gap_summary = (
    missing_df.groupby('gap_group')
    .agg(
        start=('timestamp', 'min'),
        end=('timestamp', 'max'),
        missing_hours=('timestamp', 'count')
    )
    .sort_values(
        'missing_hours',
        ascending=False
    )
)

print(gap_summary)
```

```python
missing_by_month = pd.Series(
    missing_hours.month
).value_counts().sort_index()

print(f"Missing by Month:\n{missing_by_month}")
```

```python
# For this dataset check if total rentals =  casual + registered
invalid_rentals = hourly_data[hourly_data['cnt'] != (hourly_data['casual'] + hourly_data['registered'])]

print(f"Number of invalid rental records: {len(invalid_rentals)}")
```

```python
# Expected valid ranges based on dataset documentation

valid_ranges = {
    'season': (1, 4),
    'yr': (0, 1),
    'mnth': (1, 12),
    'hr': (0, 23),
    'holiday': (0, 1),
    'weekday': (0, 6),
    'workingday': (0, 1),
    'weathersit': (1, 4),
    'temp': (0, 1),
    'atemp': (0, 1),
    'hum': (0, 1),
    'windspeed': (0, 1),
    'casual': (0, None),
    'registered': (0, None),
    'cnt': (0, None)
}

for column, (lower, upper) in valid_ranges.items():

    invalid = hourly_data[
        (hourly_data[column] < lower) |
        (hourly_data[column] > upper)
        if upper is not None
        else (hourly_data[column] < lower)
    ]

    print(f"{column}: {len(invalid)} invalid records")
```

```python
# Checking for logical mismatches inter columns
workingday_mismatch = hourly_data[(hourly_data['workingday'] == 1) & (hourly_data['holiday'] == 1)] # A working day cannot be a holiday, so this condition checks for records where both are marked as 1, indicating a logical inconsistency in the data.
print(f"Workingday and Holiday Mismatch: {len(workingday_mismatch)} records")

temperature_mismatch = hourly_data[(hourly_data['temp'] < 0) | (hourly_data['temp'] > 1)] # Temperature values should be between 0 and 1, so this condition checks for records where the temperature is outside this range, indicating a logical inconsistency in the data.
print(f"Temperature Mismatch: {len(temperature_mismatch)} records")

negative_rentals = hourly_data[(hourly_data['casual'] < 0) | (hourly_data['registered'] < 0) | (hourly_data['cnt'] < 0)] # Rental counts should not be negative, so this condition checks for records where any of the rental counts are negative, indicating a logical inconsistency in the data.
print(f"Negative Rentals: {len(negative_rentals)} records")
```

```python
#Plotting the distribution of rental counts to visualize the data and identify any potential outliers or anomalies in the dataset.
plt.figure(figsize=(12, 6))
sns.histplot(hourly_data['cnt'], bins=100, kde=True)
plt.title('Distribution of Rental Counts')
plt.xlabel('Rental Counts')
plt.ylabel('Frequency')
plt.show()
```

```python
#Boxplot for outlier detection in rental counts to visualize the spread and identify any potential outliers in the dataset.
plt.figure(figsize=(12, 6))
sns.boxplot(x=hourly_data['cnt'])
plt.title('Boxplot of Rental Counts')
plt.xlabel('Rental Counts')
plt.show()
```

```python
#IQR-based outlier detection for rental counts to identify any potential outliers in the dataset based on the interquartile range (IQR).
Q1 = hourly_data['cnt'].quantile(0.25)
Q3 = hourly_data['cnt'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR    

outliers = hourly_data[(hourly_data['cnt'] < lower_bound) | (hourly_data['cnt'] > upper_bound)]

print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
print(f"Lower Bound: {lower_bound}, Upper Bound: {upper_bound}")
print(f"Number of outliers in rental counts: {len(outliers)}")
```

```python
# Extract 20 highest-demand outliers based on rental counts to identify the top 20 records with the highest rental counts in the dataset.
print(f"Total observations: {len(hourly_data)}")
print(f"Total outliers: {len(outliers)}")

outlier_percentage = (len(outliers) / len(hourly_data)) * 100
print(f"Percentage of outliers: {outlier_percentage:.2f}%")

display(outliers.sort_values(by='cnt', ascending=False).head(20))
```

```python
#Outliers by hour

outlier_hour_distribution = (
    outliers.groupby('hr')['cnt'].agg(['count', 'mean', 'min', 'max']).sort_values(by='count', ascending=False)
)

display(outlier_hour_distribution)
```

```python
#Outliers by weathersit

outlier_weathersit_distribution = (
    outliers.groupby('weathersit')['cnt'].agg(['count', 'mean', 'min', 'max']).sort_values(by='count', ascending=False)
)

display(outlier_weathersit_distribution)
```

```python
#Outlier Rate by Weather

weather_outlier_analysis = (
    hourly_data.assign(
        is_outlier=hourly_data['cnt'] > upper_bound
    )
    .groupby('weathersit')
    .agg(
        total_observations=('cnt', 'count'),
        outlier_count=('is_outlier', 'sum'),
        average_rentals=('cnt', 'mean')
    )
)

weather_outlier_analysis['outlier_rate'] = (
    weather_outlier_analysis['outlier_count'] /
    weather_outlier_analysis['total_observations']
) * 100

display(weather_outlier_analysis)
```

```python
# Outlier Rate by Hour

hourly_outlier_analysis = (
    hourly_data.assign(
        is_outlier=hourly_data['cnt'] > upper_bound
    )
    .groupby('hr')
    .agg(
        total_observations=('cnt', 'count'),
        outlier_count=('is_outlier', 'sum'),
        average_rentals=('cnt', 'mean')
    )
)

hourly_outlier_analysis['outlier_rate'] = (
    hourly_outlier_analysis['outlier_count'] /
    hourly_outlier_analysis['total_observations']
) * 100

display(
    hourly_outlier_analysis.sort_values(
        'outlier_rate',
        ascending=False
    )
)
```

### Statistical Profiling Summary

The rental count distribution exhibits strong positive skewness,
with a concentration of observations at lower rental counts and
a long right tail extending towards higher rental counts.

The median rental count is 142, while the mean is approximately
189.46, indicating that higher rental observations influence
the overall average.

IQR-based analysis identified 505 statistical outliers,
representing approximately 2.91% of all observations.

The upper IQR boundary was calculated as 642.5 rentals.

Further investigation revealed:

- High-demand observations are concentrated around 08:00,
  17:00 and 18:00.
- Approximately 81% of identified outliers occur during
  these three hours.
- Clear weather accounts for the largest number of outliers.
- Outlier rates decrease across the observed weather
  categories from clear to light precipitation.
- Registered customers contribute substantially to the
  highest-demand observations.

The identified outliers appear to represent potentially
legitimate demand variations rather than confirmed data errors.

Therefore, no outlier removal is justified at this stage.

All observations will be retained for subsequent analysis.

Note: Weather category 4 contains only three observations,
limiting meaningful statistical interpretation of this category.
